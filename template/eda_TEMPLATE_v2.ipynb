# A. Analyses internes au dataset

**Section A — comment les variables se croisent (sans la cible)**

Dans la description, on a regardé chaque variable séparément. Ici, l'idée est de voir comment elles se croisent entre elles, ce qui pourra nous aider pour la sélection et la construction des variables.

1. **Redondances** : variables qui portent la même information (forte corrélation entre quantitatives, forte association entre qualitatives, variable calculable à partir d'autres).
2. **Variables peu informatives** : quasi constantes, ou dont l'information est entièrement contenue dans une autre. 
3. **Dépendances entre variables de nature différente** : une variable quantitative varie-t-elle selon les modalités d'une qualitative ? 
4. **Groupes et sous-populations** : des profils distincts apparaissent-ils (ex. types d'objets aux comportements très différents) ? 
5. **Incohérences entre variables** non vues en description : deux variables qui devraient être liées et ne le sont pas.
6. **Variables à construire** : ratios, regroupements, combinaisons suggérés par les croisements.

# B. Relation avec la cible

**Avant de commencer : est-ce qu'on peut rapprocher ce dataset de la cible ?**


**Les questions qu'on peut se poser :**
- **Quelles variables semblent liées à la cible**, dans quel sens, avec quelle force ?
- **Quelle forme a la relation** : linéaire, monotone, effet de seuil, rien du tout ? Ça peut orienter une transformation ou le choix des modèles.
- **La relation tient-elle partout ?** Dans tous les sous-groupes (types d'objets, catégories de territoire, périodes), ou disparaît-elle dans certains ?
- **Et dans le temps ?** Comme notre problématique porte sur l'évolution des prix, une variable liée au niveau de la cible ne l'est pas forcément à son évolution : ça vaut le coup de regarder les deux.
- **Pour les redondances vues en A**, laquelle des deux variables est la plus liée à la cible ?
- **Une variable presque parfaitement liée à la cible ?** Ça mérite qu'on vérifie qu'elle est bien disponible a priori et pas calculée à partir de la cible.

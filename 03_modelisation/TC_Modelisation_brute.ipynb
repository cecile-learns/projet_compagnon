```python
# =============================================================================
# CELLULE 1 : imports, chemins et paramètres
# =============================================================================
import sys
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.inspection import permutation_importance
from sklearn.metrics import r2_score

# config.py est à la racine du projet (comme dans preparation.ipynb)
sys.path.insert(0, str(Path.cwd().parent))
from config import DATA_OUT

lire = pd.read_parquet          # tout est lu en mémoire : aucun fichier n'est écrit

ANNEES_CIBLE = [2020, 2021, 2022, 2023, 2024]   # prix de l'année N
ANNEES_TRAIN = [2020, 2021, 2022, 2023]         # pour choisir les variables
ANNEE_VALID = 2024                              # pour mesurer le gain (jamais pour choisir)
SEUIL_VENTES = 20                               # même seuil que le plan de modélisation (5.3)


# =============================================================================
# CELLULE 2 : chargement des tables préparées (data/processed)
# =============================================================================
cible = lire(DATA_OUT / "cible_bv_annee.parquet")
liaison = lire(DATA_OUT / "liaison_commune_bv2022.parquet")
logement = lire(DATA_OUT / "logement_commune_2022.parquet")
menages = lire(DATA_OUT / "menages_commune_2022.parquet")
lovac = lire(DATA_OUT / "lovac_commune_annee.parquet")
delinq = lire(DATA_OUT / "delinquance_commune_annee.parquet")
equip = lire(DATA_OUT / "equipements_commune_millesime.parquet")
taux = lire(DATA_OUT / "taux_credit_annee.parquet")

for nom, t in [("cible", cible), ("liaison", liaison), ("logement", logement), ("menages", menages),
               ("lovac", lovac), ("delinquance", delinq), ("equipements", equip), ("taux", taux)]:
    print(nom, t.shape)


# =============================================================================
# CELLULE 3 : rattacher chaque commune à son bassin de vie
# =============================================================================
# La table de liaison contient tous les codes ayant existé depuis 2003 (et les
# arrondissements) : elle convient aux sources en géographie 2024 (recensement)
# comme en géographie 2026 (LOVAC, délinquance, équipements).
def ajouter_bassin(df, nom_source):
    out = df.merge(liaison[["CODGEO_INI", "BV2022"]], left_on="code_commune",
                   right_on="CODGEO_INI", how="left", validate="many_to_one")
    print(nom_source, ": communes sans bassin =", out.loc[out["BV2022"].isna(), "code_commune"].nunique())
    return out.drop(columns=["CODGEO_INI"])

logement = ajouter_bassin(logement, "logement")
menages = ajouter_bassin(menages, "ménages")
lovac = ajouter_bassin(lovac, "LOVAC")
delinq = ajouter_bassin(delinq, "délinquance")
equip = ajouter_bassin(equip, "équipements")


# =============================================================================
# CELLULE 4 : bloc LOGEMENT (recensement 2022) au bassin de vie
# =============================================================================
# On ADDITIONNE les effectifs des communes du bassin, puis on recalcule les taux
# (jamais de moyenne des taux communaux). Valeur fixe : la même pour toutes les années.
def ratio(numerateur, denominateur):
    return numerateur / denominateur.replace(0, np.nan) * 100

effectifs_log = [c for c in logement.columns if c.startswith("P22_") or c.startswith("C22_")]
lg = logement.groupby("BV2022")[effectifs_log].sum().reset_index()

lg["log_taux_vacance"] = ratio(lg["P22_LOGVAC"], lg["P22_LOG"])
lg["log_part_rsec"] = ratio(lg["P22_RSECOCC"], lg["P22_LOG"])
lg["log_part_maisons"] = ratio(lg["P22_MAISON"], lg["P22_LOG"])
lg["log_part_apparts"] = ratio(lg["P22_APPART"], lg["P22_LOG"])
lg["log_part_moins40m2"] = ratio(lg["P22_RP_M30M2"] + lg["P22_RP_3040M2"], lg["P22_RP"])
lg["log_part_100m2_plus"] = ratio(lg["P22_RP_100120M2"] + lg["P22_RP_120M2P"], lg["P22_RP"])
lg["log_nb_pieces_moyen"] = lg["P22_NBPI_RP"] / lg["P22_RP"]
lg["log_part_avant1946"] = ratio(lg["P22_RP_ACH1919"] + lg["P22_RP_ACH1945"], lg["P22_RP"])
lg["log_part_2006_2019"] = ratio(lg["P22_RP_ACH2019"], lg["P22_RP"])
lg["log_part_proprio"] = ratio(lg["P22_RP_PROP"], lg["P22_RP"])
lg["log_part_locataires"] = ratio(lg["P22_RP_LOC"], lg["P22_RP"])
lg["log_part_hlm"] = ratio(lg["P22_RP_LOCHLMV"], lg["P22_RP"])
lg["log_part_emmenages_2ans"] = ratio(lg["P22_MEN_ANEM0002"], lg["P22_MEN"])
lg["log_anciennete_moyenne"] = lg["P22_ANEM_RP"] / lg["P22_RP"]
lg["log_taille_menage"] = lg["P22_NPER_RP"] / lg["P22_RP"]
lg["log_part_sans_voiture"] = 100 - ratio(lg["P22_RP_VOIT1P"], lg["P22_MEN"])
lg["log_part_2voitures"] = ratio(lg["P22_RP_VOIT2P"], lg["P22_MEN"])
lg["log_part_parking"] = ratio(lg["P22_RP_GARL"], lg["P22_MEN"])
lg["log_part_suroccupes"] = ratio(lg["C22_RP_SUROCC_MOD"] + lg["C22_RP_SUROCC_ACC"], lg["P22_RP"])
lg["log_nb_logements"] = np.log10(lg["P22_LOG"])          # taille du bassin

vars_logement = [c for c in lg.columns if c.startswith("log_")]
lg = lg[["BV2022"] + vars_logement]
print("Logement au bassin :", lg.shape)


# =============================================================================
# CELLULE 5 : bloc LOGEMENT (suite) : LOVAC au bassin, par année
# =============================================================================
# Règle N-1 : le prix de l'année N reçoit le millésime N, qui décrit la situation
# au 1er janvier N-1 (colonne annee_vente_associee).
# Seules les communes dont les effectifs sont connus (hors secret) sont additionnées.
lv = lovac[lovac["annee_vente_associee"].isin(ANNEES_CIBLE)].copy()

connu = lv["vacants"].notna() & lv["parc_prive"].notna()
s1 = lv[connu].groupby(["BV2022", "annee_vente_associee"])[["vacants", "parc_prive"]].sum()

connu2 = lv["vacants_2ans"].notna() & lv["parc_prive"].notna()
s2 = lv[connu2].groupby(["BV2022", "annee_vente_associee"])[["vacants_2ans", "parc_prive"]].sum()
s2.columns = ["vacants_2ans", "parc_prive_2"]

connu3 = lv["vacants_2ans"].notna() & lv["vacants"].notna()
s3 = lv[connu3].groupby(["BV2022", "annee_vente_associee"])[["vacants_2ans", "vacants"]].sum()
s3.columns = ["vac2_b", "vac_b"]

# Part des communes du bassin dont le parc privé est connu (fiabilité du taux)
couv = connu.groupby([lv["BV2022"], lv["annee_vente_associee"]]).mean() * 100
couv.name = "lovac_couverture"

cles = ["BV2022", "annee_vente_associee"]
lb = s1.reset_index().merge(s2.reset_index(), on=cles, how="outer")
lb = lb.merge(s3.reset_index(), on=cles, how="outer")
lb = lb.merge(couv.reset_index(), on=cles, how="outer")
lb["lovac_taux_vacance"] = ratio(lb["vacants"], lb["parc_prive"])
lb["lovac_taux_vacance_2ans"] = ratio(lb["vacants_2ans"], lb["parc_prive_2"])
lb["lovac_part_longue"] = ratio(lb["vac2_b"], lb["vac_b"])
lb = lb.rename(columns={"annee_vente_associee": "annee"})
vars_lovac = ["lovac_taux_vacance", "lovac_taux_vacance_2ans", "lovac_part_longue"]
lb = lb[["BV2022", "annee"] + vars_lovac + ["lovac_couverture"]]
print("LOVAC au bassin :", lb.shape)
print(lb[vars_lovac + ["lovac_couverture"]].describe().round(1))


# =============================================================================
# CELLULE 6 : bloc DÉMOGRAPHIE (ménages, recensement 2022) au bassin
# =============================================================================
eff_men = ["C22_MEN", "C22_PMEN", "C22_MENPSEUL", "C22_MENFAMMONO", "C22_NE24F3", "C22_NE24F4P", "C22_MENCOUPSENF"]
mn = menages.groupby("BV2022")[eff_men].sum().reset_index()
mn["dem_part_personnes_seules"] = ratio(mn["C22_MENPSEUL"], mn["C22_MEN"])
mn["dem_part_monoparentales"] = ratio(mn["C22_MENFAMMONO"], mn["C22_MEN"])
mn["dem_part_familles_nombreuses"] = ratio(mn["C22_NE24F3"] + mn["C22_NE24F4P"], mn["C22_MEN"])
mn["dem_taille_menage"] = mn["C22_PMEN"] / mn["C22_MEN"]
mn["dem_part_couples_sans_enfant"] = ratio(mn["C22_MENCOUPSENF"], mn["C22_MEN"])
vars_demo = [c for c in mn.columns if c.startswith("dem_")]
mn = mn[["BV2022"] + vars_demo]


# =============================================================================
# CELLULE 7 : bloc ATTRACTIVITÉ (délinquance N-1, équipements) au bassin
# =============================================================================
# Délinquance : nombres de faits et population additionnés, puis taux pour 1 000
# habitants (cambriolages pour 1 000 logements). Année N-1 -> prix de l'année N.
colonnes_nb = [c for c in delinq.columns if c.startswith("nb_")]
dl = delinq.groupby(["BV2022", "annee"])[colonnes_nb + ["insee_pop", "insee_log"]].sum().reset_index()
for col in colonnes_nb:
    if col == "nb_cambriolages_de_logement":
        dl["att_" + col[3:]] = dl[col] / dl["insee_log"].replace(0, np.nan) * 1000
    else:
        dl["att_" + col[3:]] = dl[col] / dl["insee_pop"].replace(0, np.nan) * 1000
dl["annee"] = dl["annee"] + 1                     # décalage : faits de N-1 pour le prix de N
vars_delinq = [c for c in dl.columns if c.startswith("att_")]
dl = dl[["BV2022", "annee"] + vars_delinq]

# Équipements : indicateurs de présence (1/0/inconnu). Au bassin : nombre de types
# d'équipements présents dans au moins une commune du bassin.
# Millésime connu en N-1 : BPE 2015 pour les prix 2020 (N-1 = 2019), BPE 2020 ensuite.
col_bpe = [c for c in equip.columns if c.startswith("bpe_presence_")]
eq = equip.groupby(["BV2022", "millesime_bpe"])[col_bpe].max()      # 1 si présent dans une commune
eq["att_nb_types_equipements"] = (eq == 1).sum(axis=1)
eq = eq[["att_nb_types_equipements"]].reset_index()
corresp = pd.DataFrame({"annee": ANNEES_CIBLE, "millesime_bpe": [2015, 2020, 2020, 2020, 2020]})
eq = corresp.merge(eq, on="millesime_bpe").drop(columns="millesime_bpe")
vars_attract = vars_delinq + ["att_nb_types_equipements"]


# =============================================================================
# CELLULE 8 : TEMPS : taux des crédits de l'année N-1
# =============================================================================
tx = taux.copy()
tx["annee"] = tx["annee"] + 1
tx = tx.rename(columns={"taux": "taux_n1"})


# =============================================================================
# CELLULE 9 : la table cible (une ligne = bassin × année × type de bien)
# =============================================================================
morceaux = []
for t in ["maison", "appart"]:
    c = cible[["BV2022", "annee", "n_" + t, "prix_m2_med_" + t]].copy()
    c.columns = ["BV2022", "annee", "nb_ventes", "prix_m2"]
    c["type_bien"] = t
    # Prix de l'année N-1, seulement si le bassin avait au moins 20 ventes en N-1
    prec = c[c["nb_ventes"] >= SEUIL_VENTES][["BV2022", "annee", "prix_m2"]].copy()
    prec["annee"] = prec["annee"] + 1
    prec = prec.rename(columns={"prix_m2": "prix_m2_n1"})
    c = c.merge(prec, on=["BV2022", "annee"], how="left")
    # Lignes gardées : années 2020-2024 et au moins 20 ventes l'année N
    c = c[c["annee"].isin(ANNEES_CIBLE) & (c["nb_ventes"] >= SEUIL_VENTES)]
    morceaux.append(c)
table = pd.concat(morceaux, ignore_index=True)
print("Table cible :", table.shape)
print("Lignes sans prix N-1 :")
print(table[table["prix_m2_n1"].isna()].groupby("type_bien").size())


# =============================================================================
# CELLULE 10 : jointure de tous les blocs et contrôles
# =============================================================================
n_avant = len(table)
table = table.merge(tx, on="annee", how="left")
table = table.merge(lg, on="BV2022", how="left")
table = table.merge(lb, on=["BV2022", "annee"], how="left")
table = table.merge(mn, on="BV2022", how="left")
table = table.merge(dl, on=["BV2022", "annee"], how="left")
table = table.merge(eq, on=["BV2022", "annee"], how="left")

print("Lignes avant / après jointures :", n_avant, "/", len(table))
print("Une ligne par bassin × année × type :",
      not table.duplicated(["BV2022", "annee", "type_bien"]).any())
print("Table au bassin :", table.shape)
toutes_vars = ["taux_n1"] + vars_logement + vars_lovac + vars_demo + vars_attract
manq = table[toutes_vars].isna().mean() * 100
print("Variables avec des valeurs manquantes (%) :")
print(manq[manq > 0].round(1))

# Valeurs du taux N-1 vues par le modèle, par année
print(table.groupby("annee")["taux_n1"].first())


# =============================================================================
# CELLULE 11 : la référence naïve (prix N = prix N-1), chiffrée sur 2024
# =============================================================================
# On garde les lignes qui ont un prix N-1 (sinon aucune prévision possible).
# nb_ventes est celui de l'année N : il sert au seuil mais n'entre jamais dans un modèle.
t = table[table["prix_m2_n1"].notna()].copy()
t["erreur_naive"] = (t["prix_m2_n1"] - t["prix_m2"]).abs()
t["erreur_naive_pct"] = t["erreur_naive"] / t["prix_m2"] * 100
naive = t.groupby(["type_bien", "annee"])[["erreur_naive", "erreur_naive_pct"]].mean().round(1)
naive.columns = ["MAE (€/m²)", "erreur moyenne (%)"]
print(naive)


# =============================================================================
# CELLULE 12 : corrélations de Spearman (années d'apprentissage seulement)
# =============================================================================
# Deux cibles : le NIVEAU (log du prix N) et l'ÉVOLUTION (log prix N - log prix N-1).
t["log_prix"] = np.log(t["prix_m2"])
t["evolution"] = np.log(t["prix_m2"]) - np.log(t["prix_m2_n1"])
train = t[t["annee"].isin(ANNEES_TRAIN)]
valid = t[t["annee"] == ANNEE_VALID]

def spearman(df, x, y):
    ok = df[x].notna() & df[y].notna()
    return stats.spearmanr(df.loc[ok, x], df.loc[ok, y])[0]

lignes = []
for v in toutes_vars:
    ligne = {"variable": v}
    for tb in ["maison", "appart"]:
        d = train[train["type_bien"] == tb]
        ligne["niveau_" + tb] = spearman(d, v, "log_prix")
        ligne["evol_" + tb] = spearman(d, v, "evolution")
    lignes.append(ligne)
corr = pd.DataFrame(lignes).set_index("variable").round(2)
corr["force_niveau"] = corr[["niveau_maison", "niveau_appart"]].abs().max(axis=1)
corr["force_evol"] = corr[["evol_maison", "evol_appart"]].abs().max(axis=1)
corr = corr.sort_values("force_niveau", ascending=False)
display(corr)

# Doublons : paires de variables corrélées à plus de 0,8 entre elles
m = train[toutes_vars].corr(method="spearman")
print("Paires en doublon (|rho| > 0,8) :")
for i in range(len(toutes_vars)):
    for j in range(i + 1, len(toutes_vars)):
        a = toutes_vars[i]
        b = toutes_vars[j]
        if abs(m.loc[a, b]) > 0.8:
            print("  ", a, "/", b, ":", round(m.loc[a, b], 2))

# Graphique 1 : lien de chaque variable avec le niveau et l'évolution du prix
fig, axes = plt.subplots(1, 2, figsize=(14, 11), sharey=True)
ordre = corr.index[::-1]
for ax, tb, titre in [(axes[0], "maison", "Maisons"), (axes[1], "appart", "Appartements")]:
    y = np.arange(len(ordre))
    ax.barh(y - 0.2, corr.loc[ordre, "niveau_" + tb], height=0.4, label="niveau du prix N")
    ax.barh(y + 0.2, corr.loc[ordre, "evol_" + tb], height=0.4, label="évolution N-1 -> N")
    ax.set_yticks(y)
    ax.set_yticklabels(ordre)
    ax.axvline(0, color="grey", lw=0.8)
    ax.axvline(0.2, color="grey", ls=":")
    ax.axvline(-0.2, color="grey", ls=":")
    ax.set_title(titre + " : corrélation de Spearman (2020-2023)")
    ax.set_xlabel("rho")
axes[0].legend(loc="lower left")
plt.tight_layout()
plt.show()


# =============================================================================
# CELLULE 13 : ÉVOLUTION du prix : peut-on battre la prévision naïve ?
# =============================================================================
# Le modèle prédit l'évolution (log prix N - log prix N-1) ; le prix prédit vaut
# prix N-1 × exp(évolution prédite). Il bat la naïve seulement s'il prédit mieux
# l'évolution que « aucune évolution ».
# Valeurs manquantes : médiane de l'apprentissage. Standardisation : moyenne et
# écart-type de l'apprentissage. Rien n'est calculé sur 2024.
def preparer(tr, va, variables):
    med = tr[variables].median()
    moy = tr[variables].fillna(med).mean()
    ect = tr[variables].fillna(med).std().replace(0, 1)
    xtr = (tr[variables].fillna(med) - moy) / ect
    xva = (va[variables].fillna(med) - moy) / ect
    return xtr, xva

def arbres():
    return HistGradientBoostingRegressor(max_iter=300, learning_rate=0.05, random_state=0)

# Graphique 2 : évolution moyenne des prix et taux N-1, par année
evol = t.groupby(["annee", "type_bien"])["evolution"].mean().unstack() * 100
fig, ax = plt.subplots(figsize=(9, 4.5))
evol.plot(ax=ax, marker="o")
ax.axhline(0, color="grey", lw=0.8)
ax.axvspan(2023.5, 2024.5, color="orange", alpha=0.15)
ax.set_ylabel("évolution moyenne du prix N-1 -> N (%)")
ax.set_xlabel("année N (zone orange : validation)")
ax2 = ax.twinx()
ax2.plot(tx[tx["annee"].isin(ANNEES_CIBLE)]["annee"], tx[tx["annee"].isin(ANNEES_CIBLE)]["taux_n1"],
         color="black", ls="--", marker="s", label="taux N-1")
ax2.set_ylabel("taux des crédits N-1 (%)")
ax2.legend(loc="lower left")
ax.set_title("Apprentissage en hausse, validation en baisse, taux hors de la plage apprise")
plt.tight_layout()
plt.show()

res_evol = []
for tb in ["maison", "appart"]:
    tr = train[train["type_bien"] == tb]
    va = valid[valid["type_bien"] == tb]
    err_naive = (va["prix_m2_n1"] - va["prix_m2"]).abs()
    # Naïve + hausse moyenne de l'apprentissage
    err_tend = (va["prix_m2_n1"] * np.exp(tr["evolution"].mean()) - va["prix_m2"]).abs()
    res_evol.append({"type": tb, "modèle": "naïve", "MAE (€/m²)": err_naive.mean()})
    res_evol.append({"type": tb, "modèle": "naïve + hausse moyenne 2020-2023", "MAE (€/m²)": err_tend.mean()})
    for nom, variables in [("territoire seul", vars_logement + vars_lovac + vars_demo + vars_attract),
                           ("territoire + taux N-1", toutes_vars)]:
        xtr, xva = preparer(tr, va, variables)
        lin = LinearRegression().fit(xtr, tr["evolution"])
        arb = arbres().fit(tr[variables], tr["evolution"])
        for algo, pred in [("linéaire", lin.predict(xva)), ("arbres", arb.predict(va[variables]))]:
            err = (va["prix_m2_n1"] * np.exp(pred) - va["prix_m2"]).abs()
            res_evol.append({"type": tb, "modèle": nom + " | " + algo, "MAE (€/m²)": err.mean()})
res_evol = pd.DataFrame(res_evol)
res_evol["écart à la naïve (€/m²)"] = res_evol["MAE (€/m²)"] - res_evol.groupby("type")["MAE (€/m²)"].transform("first")
display(res_evol.round(1))


# =============================================================================
# CELLULE 14 : NIVEAU du prix : ce que chaque bloc apporte
# =============================================================================
# Ici, pas de prix N-1 : on mesure ce que les variables du territoire expliquent
# des ÉCARTS DE PRIX entre bassins. On ajoute les blocs un par un (MCD).

# 1. Doublons retirés à l'intérieur de chaque bloc (on garde la variable la plus
#    liée au niveau du prix). log_taille_menage est retirée : c'est la même que
#    dem_taille_menage (rho = 1), qui appartient au bloc démographie.
force = corr["force_niveau"]

def dedoublonner(variables):
    gardees = list(variables)
    for i in range(len(variables)):
        for j in range(i + 1, len(variables)):
            a = variables[i]
            b = variables[j]
            if abs(m.loc[a, b]) > 0.8 and a in gardees and b in gardees:
                if force[a] < force[b]:
                    gardees.remove(a)
                else:
                    gardees.remove(b)
    return gardees

bloc_log = dedoublonner([v for v in vars_logement + vars_lovac if v != "log_taille_menage"])
bloc_dem = dedoublonner(vars_demo)
bloc_att = dedoublonner(vars_attract)
print("Logement :", bloc_log)
print("Démographie :", bloc_dem)
print("Attractivité :", bloc_att)

etapes = [("logement", bloc_log),
          ("+ démographie", bloc_log + bloc_dem),
          ("+ attractivité", bloc_log + bloc_dem + bloc_att)]

res_niv = []
for tb in ["maison", "appart"]:
    tr = train[train["type_bien"] == tb]
    va = valid[valid["type_bien"] == tb]
    precedent = None
    for nom, variables in etapes:
        xtr, xva = preparer(tr, va, variables)
        lin = LinearRegression().fit(xtr, tr["log_prix"])
        arb = arbres().fit(tr[variables], tr["log_prix"])
        p_lin = lin.predict(xva)
        p_arb = arb.predict(va[variables])
        err_lin = (np.exp(p_lin) - va["prix_m2"]).abs()
        ligne = {"type": tb, "blocs": nom, "nb variables": len(variables),
                 "MAE linéaire (€/m²)": err_lin.mean(),
                 "R² linéaire": r2_score(va["log_prix"], p_lin),
                 "MAE arbres (€/m²)": (np.exp(p_arb) - va["prix_m2"]).abs().mean(),
                 "R² arbres": r2_score(va["log_prix"], p_arb)}
        # Test apparié (t de Student) : le bloc fait-il baisser l'erreur de la régression ?
        if precedent is not None:
            diff = precedent - err_lin.values
            ligne["gain du bloc (€/m²)"] = diff.mean()
            ligne["p-value"] = stats.ttest_rel(precedent, err_lin.values)[1]
            ligne["d de Cohen"] = diff.mean() / diff.std()
        precedent = err_lin.values
        res_niv.append(ligne)
res_niv = pd.DataFrame(res_niv)
display(res_niv.round(3))

# Graphique 3 : erreur sur 2024 selon les blocs, comparée à la naïve
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5), sharey=True)
for ax, tb, titre in [(axes[0], "maison", "Maisons"), (axes[1], "appart", "Appartements")]:
    r = res_niv[res_niv["type"] == tb]
    x = np.arange(len(r))
    ax.bar(x - 0.2, r["MAE linéaire (€/m²)"], width=0.4, label="régression linéaire")
    ax.bar(x + 0.2, r["MAE arbres (€/m²)"], width=0.4, label="arbres")
    naive_tb = (valid[valid["type_bien"] == tb]["prix_m2_n1"] - valid[valid["type_bien"] == tb]["prix_m2"]).abs().mean()
    ax.axhline(naive_tb, color="black", ls="--", label="naïve (prix N-1)")
    ax.set_xticks(x)
    ax.set_xticklabels(r["blocs"])
    ax.set_title(titre + " : erreur sur 2024, sans prix N-1")
    ax.set_ylabel("MAE (€/m²)")
axes[0].legend()
plt.tight_layout()
plt.show()


# =============================================================================
# CELLULE 15 : importance des variables (SHAP linéaire et permutation)
# =============================================================================
# SHAP d'une régression linéaire sur variables standardisées : coefficient × valeur
# (c'est ce que calcule shap.LinearExplainer). On l'exprime en % du total.
# Contrôle avec les arbres : importance par permutation (baisse de qualité quand on
# mélange une variable), aussi en %.
toutes_gardees = bloc_log + bloc_dem + bloc_att
imp = {}
for tb in ["maison", "appart"]:
    tr = train[train["type_bien"] == tb]
    va = valid[valid["type_bien"] == tb]
    xtr, xva = preparer(tr, va, toutes_gardees)
    lin = LinearRegression().fit(xtr, tr["log_prix"])
    shap = np.abs(xva.values * lin.coef_).mean(axis=0)
    arb = arbres().fit(tr[toutes_gardees], tr["log_prix"])
    perm = permutation_importance(arb, va[toutes_gardees], va["log_prix"], n_repeats=5,
                                  random_state=0, scoring="neg_mean_absolute_error").importances_mean
    perm = np.clip(perm, 0, None)
    imp[tb] = pd.DataFrame({"shap_%": shap / shap.sum() * 100,
                            "permutation_%": perm / perm.sum() * 100,
                            "sens": np.sign(lin.coef_)}, index=toutes_gardees)

# Graphique 4 : importance des variables, maisons et appartements
fig, axes = plt.subplots(1, 2, figsize=(16, 10))
for ax, tb, titre in [(axes[0], "maison", "Maisons"), (axes[1], "appart", "Appartements")]:
    d = imp[tb].sort_values("shap_%")
    y = np.arange(len(d))
    ax.barh(y - 0.2, d["shap_%"], height=0.4, label="|SHAP| moyen, régression (%)")
    ax.barh(y + 0.2, d["permutation_%"], height=0.4, label="permutation, arbres (%)")
    ax.set_yticks(y)
    ax.set_yticklabels(d.index)
    ax.axvline(2, color="grey", ls=":")
    ax.set_title(titre + " : importance (niveau, 2024)")
axes[0].legend(loc="lower right")
plt.tight_layout()
plt.show()


# =============================================================================
# CELLULE 16 : décision par variable
# =============================================================================
# Règle : garder si |rho| > 0,2 avec le niveau du prix (maisons ou appartements)
# ET importance >= 2 % (SHAP ou permutation, maisons ou appartements).
dec = corr.loc[toutes_gardees, ["niveau_maison", "niveau_appart", "evol_maison", "evol_appart"]].copy()
for tb in ["maison", "appart"]:
    dec["shap_" + tb] = imp[tb]["shap_%"]
    dec["perm_" + tb] = imp[tb]["permutation_%"]
lien = dec[["niveau_maison", "niveau_appart"]].abs().max(axis=1) > 0.2
importance = dec[["shap_maison", "shap_appart", "perm_maison", "perm_appart"]].max(axis=1) >= 2
dec["décision"] = np.where(lien & importance, "garder", "écarter")
dec = dec.sort_values(["décision", "niveau_maison"], ascending=[False, True])
display(dec.round(2))
print("Variables écartées comme doublons :",
      [v for v in vars_logement + vars_lovac + vars_demo + vars_attract if v not in toutes_gardees])

# =============================================================================
# CELLULE 17 : distribution des valeurs SHAP (style shap.summary_plot), validation 2024
# =============================================================================
# Un point = un bassin de 2024. Position horizontale = valeur SHAP de la variable
# pour ce bassin (effet sur le log du prix : 0,1 ≈ +10 %).
# Couleur = valeur de la variable (bleu = faible, rouge = élevée).
# Les points s'empilent verticalement là où ils sont nombreux (densité).
# Variables triées selon la somme des |SHAP| (la plus forte en haut).
# Ligne rouge : sous la ligne, la variable pèse moins de 2 % du total des |SHAP| :
# elle est candidate à la suppression.
from matplotlib.colors import LinearSegmentedColormap

SEUIL_IMPORTANCE = 2      # en % du total des |SHAP|
couleurs_shap = LinearSegmentedColormap.from_list("shap", ["#008bfb", "#ff0052"])   # palette du cours

def decalage_densite(valeurs, largeur=0.4, nb_classes=60):
    """Décalage vertical de chaque point, proportionnel au nombre de points voisins."""
    classes = np.linspace(valeurs.min(), valeurs.max() + 1e-9, nb_classes + 1)
    num = np.digitize(valeurs, classes) - 1                     # classe de chaque point
    effectif = np.bincount(num, minlength=nb_classes + 1)        # nombre de points par classe
    hauteur = effectif[num] / effectif.max() * largeur            # nuage plus épais là où c'est dense
    rng = np.random.default_rng(0)
    return rng.uniform(-1, 1, len(valeurs)) * hauteur

for tb, titre in [("maison", "Maisons"), ("appart", "Appartements")]:
    tr = train[train["type_bien"] == tb]
    va = valid[valid["type_bien"] == tb]
    xtr, xva = preparer(tr, va, toutes_gardees)
    lin = LinearRegression().fit(xtr, tr["log_prix"])
    shap = pd.DataFrame(xva.values * lin.coef_, columns=toutes_gardees)   # SHAP de la régression

    # Part de chaque variable dans le total des |SHAP| (en %), de la plus forte à la plus faible
    part = shap.abs().sum() / shap.abs().sum().sum() * 100
    part = part.sort_values(ascending=False)
    ordre = list(part.index)
    nb_importantes = (part >= SEUIL_IMPORTANCE).sum()

    fig, ax = plt.subplots(figsize=(10, 12))
    n = len(ordre)
    for rang, v in enumerate(ordre):
        k = n - 1 - rang                                 # la plus importante en haut
        x = shap[v].values
        y = k + decalage_densite(x)
        couleur = va[v].rank(pct=True).values           # rang de la valeur entre 0 et 1
        points = ax.scatter(x, y, c=couleur, cmap=couleurs_shap, s=8, alpha=0.8,
                            vmin=0, vmax=1, linewidths=0)
        ax.axhline(k, color="#cccccc", lw=0.5, ls=":", zorder=0)
    # Étiquettes : nom de la variable et sa part en %, en gris sous le seuil
    etiquettes = []
    for v in ordre[::-1]:
        etiquettes.append(v + "  (" + str(round(part[v], 1)) + " %)")
    ax.set_yticks(range(n))
    ax.set_yticklabels(etiquettes)
    for rang, lab in enumerate(ax.get_yticklabels()[::-1]):
        if rang >= nb_importantes:
            lab.set_color("#999999")

    # Barre de séparation : au-dessus, variables importantes ; en dessous, supprimables
    y_seuil = n - nb_importantes - 0.5
    ax.axhline(y_seuil, color="red", lw=1.5, ls="--",
               label="seuil de 2 % : au-dessus, variables utiles ;\nen dessous (noms en gris), variables supprimables")
    ax.legend(loc="lower right", fontsize=9, frameon=True)

    ax.axvline(0, color="#999999", lw=1)
    ax.set_xlabel("valeur SHAP (impact sur le log du prix prédit ; 0,1 ≈ +10 %)")
    ax.set_title(titre + " : valeurs SHAP (niveau du prix, 2024)", loc="left")
    for cote in ["top", "right", "left"]:
        ax.spines[cote].set_visible(False)
    ax.tick_params(axis="y", length=0)
    barre = fig.colorbar(points, ax=ax, ticks=[0, 1], aspect=50)
    barre.ax.set_yticklabels(["Faible", "Élevée"])
    barre.set_label("Valeur de la variable")
    barre.outline.set_visible(False)
    plt.tight_layout()
    plt.show()
    print(titre, ": variables supprimables ->", ordre[nb_importantes:])
```

## Exploration naïve sur la table au bassin de vie (2020-2024)

### Choix de méthode
| Élément | Choix | Pourquoi |
|---|---|---|
| Table | Une ligne = bassin de vie × année × type de bien (prix 2020-2024) | Une seule table de référence pour tout le groupe |
| Seuil | 20 ventes l'année N (et l'année N-1 pour le prix N-1) | Médiane fiable (plan 5.3) |
| Agrégation au bassin | Effectifs additionnés, puis taux recalculés | Une moyenne de taux donnerait le même poids à un village et à une ville |
| Règle des années | Prix, taux, LOVAC et délinquance de N-1 pour le prix de N ; recensement 2022 en valeur fixe ; BPE 2015 pour 2020, BPE 2020 ensuite | Pas de fuite : on n'utilise que le passé |
| Années | Choix des variables sur 2020-2023 ; 2024 sert seulement à mesurer le gain | 2024 est l'année de validation |
| Mesures | MAE en €/m² et erreur en % ; R² sur le log du prix | Lisibles, demandées par le mentor |
| Modèles | Régression linéaire (variables standardisées) et gradient boosting (arbres) | Le plus simple, puis un modèle non linéaire |
| Tests | Spearman ; t de Student apparié + d de Cohen pour comparer deux modèles sur les mêmes bassins | Savoir si un gain n'est pas dû au hasard |
| Importance | SHAP de la régression (coefficient × valeur standardisée) et permutation sur les arbres | Deux méthodes pour ne pas dépendre d'un seul modèle |

### La table
- **11 892 lignes × 52 colonnes** : 8 042 lignes maisons (environ 1 609 bassins par an) et 3 850 lignes appartements (724 à 810 bassins par an).
- Aucune ligne perdue ni dupliquée par les jointures. Seule une commune LOVAC n'a pas de bassin.
- **Prix N-1 manquant** (bassin sous 20 ventes l'année d'avant) : 196 lignes appartements, 3 lignes maisons. Elles sont retirées des modèles.
- `nb_ventes` est celui de l'année N : il sert au seuil, mais n'entre dans aucun modèle (fuite).
- **taux_n1 par année : 1,36 (2020), 1,26 (2021), 1,15 (2022), 1,47 (2023), 3,02 (2024).**

|---|---|---|
| Maisons | 124,5 | 6,0 % |
| Appartements | 132,8 | 5,6 % |

### Résultat 1 : l'évolution des prix n'est pas prévisible avec ces données
| Modèle (évolution N-1 → N) | Maisons MAE 2024 | Apparts MAE 2024 |
|---|---|---|
| Naïve | **124,5** | **132,8** |
| Naïve + hausse moyenne 2020-2023 | 207,0 | 211,4 |
| Territoire seul, linéaire / arbres | 200,5 / 192,1 | 205,9 / 197,2 |
| Territoire + taux N-1, linéaire / arbres | 605,4 / 152,0 | 543,2 / 175,1 |

**Constat :** aucun modèle ne bat la naïve. Les quatre années d'apprentissage sont en hausse (+4 à +9 % par an), alors que 2024 est en baisse (-2,5 % pour les maisons, -0,7 % pour les appartements). Les modèles ont donc appris que « les prix montent ». Le taux de crédit est la variable la plus liée à l'évolution (ρ = -0,44 maisons, -0,30 appartements), mais le modèle n'en voit que **4 valeurs**, entre 1,15 et 1,47 %. En 2024, il vaut **3,02 %**, hors de tout ce qui a été appris :
- la régression linéaire **prolonge sa droite** et prédit une chute de prix énorme (erreur multipliée par 5) ;
- les arbres **ne savent pas extrapoler** : ils traitent 3,02 % comme 1,47 %, la plus haute valeur vue, et prédisent l'évolution de 2023.

Les variables du territoire sont presque sans lien avec l'évolution (|ρ| < 0,16).

**Avis métier :** l'évolution d'une année dépend surtout du **moment du marché** (taux, contexte économique), qui est le même pour tous les bassins. Avec cinq années et un seul retournement, aucun modèle ne peut apprendre comment le marché réagit à une forte remontée des taux.

### Résultat 2 : les écarts de prix entre bassins s'expliquent par le logement
| Blocs (sans prix N-1) | Maisons : linéaire MAE / R² | Maisons : arbres MAE / R² | Apparts : linéaire MAE / R² | Apparts : arbres MAE / R² |
|---|---|---|---|---|
| Logement (15 variables) | 322 / 0,84 | 151 / 0,96 | 498 / 0,76 | 256 / 0,94 |
| + démographie | 320 / 0,84 | 148 / 0,96 | 496 / 0,77 | 252 / 0,94 |
| + attractivité | 316 / 0,85 | 174 / 0,95 | 471 / 0,79 | 216 / 0,95 |

- **Démographie :** gain de 2,3 €/m² pour les maisons (p = 0,13 ; d = 0,04) et de 1,6 €/m² pour les appartements (p = 0,51) : **pas d'apport**. L'information est déjà portée par le bloc logement.
- **Attractivité :** pas de gain pour les maisons (p = 0,52 ; les arbres font même moins bien). Gain de 25 €/m² pour les appartements (p = 0,02 ; d = 0,09) : il est significatif, mais l'effet est négligeable.

**Constat :** le bloc logement explique à lui seul 84 % (linéaire) à 96 % (arbres) des écarts de log-prix entre bassins pour les maisons. Les variables les plus importantes, avec les deux méthodes, sont le **taux de vacance**, l'**âge du parc** (avant 1946, construit 2006-2019), l'**ancienneté d'emménagement**, la **part de maisons**, la **vacance longue (LOVAC)**, la **part de résidences secondaires** et la **suroccupation** (cette dernière surtout pour les appartements).

### Impact projet
- **La cible doit être tranchée :**
  - en **niveau**, avec le prix N-1, le score sera bon de toute façon : seul compte l'écart avec la naïve, et nous ne le battons pas encore ;
  - en **évolution**, le problème principal est le changement de régime de marché, pas les variables du territoire.
- **Avec seulement 2020-2024, le modèle n'a vu aucune baisse.** C'est un argument fort pour **revenir à 2015-2023** (plan 5.2), qui contient au moins une phase de stabilité, même si certaines sources y deviennent des valeurs fixes.
- **Le taux est la clé de l'évolution,** mais avec une valeur par an, il agit comme un effet « année ». Le volet B (séries temporelles) est mieux placé pour l'étudier.

### Actions envisagées
1. Refaire la même analyse sur 2015-2023 (avec LOVAC 2023 et la délinquance moyenne en valeurs fixes) et comparer.
2. Pour le taux, tester une forme qui extrapole correctement : la **variation** du taux sur un an (Δtaux) plutôt que son niveau, ou une régression linéaire contrainte sur cette seule variable.
3. Garder le bloc logement (14 variables) pour l'étape « ajout des sources » ; tester l'attractivité seulement pour les appartements.
4. Écrire noir sur blanc la cible retenue (niveau ou évolution) et toujours rapporter l'écart avec la naïve.

### Incertitudes et limites
- Les SHAP de la régression sont instables quand deux variables se ressemblent : `att_nb_types_equipements` a un SHAP élevé, mais elle est très liée à `log_nb_logements` (ρ = 0,84) et leurs coefficients se compensent. La permutation sur les arbres lui donne moins de 1 %.
- Le recensement 2022 est postérieur aux prix 2020-2022 (anticipation) et n'a été publié qu'en 2025.
- LOVAC au bassin n'additionne que les communes hors secret : en médiane, la moitié des communes d'un bassin. Le taux reflète surtout les plus grandes communes (colonne `lovac_couverture`).
- Les mêmes bassins se répètent sur plusieurs années : les p-values des corrélations sont trop optimistes. On conclut sur les tailles d'effet.
- Le bassin de Paris regroupe 440 communes : le modèle ne dit rien d'un arrondissement en particulier.

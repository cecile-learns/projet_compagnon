### 5.1 Valeurs manquantes

**Hypothèse de mécanisme, si investiguée :** …

**Éléments à l’appui et limites :** …

MCAR : absence indépendante des valeurs observées et non observées. MAR : à valeurs observées données, l’absence ne dépend pas des valeurs manquantes. MNAR : une dépendance aux valeurs non observées subsiste. Le taux de manquants ou une association observée ne suffit pas à établir le mécanisme.

```python
#Vérification des quantité de valeur manquante dans le dataset
missing = (df.isna().mean() * 100).round(2)
missing = missing[missing > 0].sort_values(ascending=False)

display(missing)
```

```python
#R2partition des valeurs manquantes en fonction du type de biens
colonnes = missing[missing.between(25, 80)].index

df_missing = df.loc[df["typedebien_lite"] != "l", colonnes]

resultat = (
    df_missing.isna()
    .groupby(df.loc[df_missing.index, "typedebien_lite"])
    .mean() * 100
).round(1)

display(resultat.T)
```

```python
display(pd.crosstab(
    df["logement_neuf"],
    df["annee_construction"].isna(),
    normalize="index"
).mul(100).round(1))
print("60% des logements dans l'ancien, ont leur année de construction manquante")
```

```python
#Répartition des catégorie d'annonceurs
repartition = pd.DataFrame({
    "Effectif": df["categorie_annonceur"].value_counts(),
    "Proportion (%)": (
        df["categorie_annonceur"].value_counts(normalize=True) * 100
    ).round(2)
})

display(repartition)

# Taux de valeurs manquantes selon la catégorie d'annonceur
colonnes = missing[missing.between(25, 80)].index

resultat_annonceur = (
    df[colonnes].isna()
    .groupby(df["categorie_annonceur"])
    .mean() * 100
).round(1)

display(resultat_annonceur.T)
```

```python
pd.crosstab(
    df["categorie_annonceur"],
    df["annee_construction"].isna(),
    normalize="index"
).mul(100).round(1)
```

```python
#Vérification de l'évolution des valeurs manquantes dans le temps
dates = pd.to_datetime(df["date"], errors="coerce")

resultat_date = (
    df["annee_construction"].isna()
    .groupby(dates.dt.year)
    .agg(["mean", "size"])
)

resultat_date["mean"] = (resultat_date["mean"] * 100).round(1)

display(resultat_date)
```

```python
#Vérification des absences groupées n6
colonnes_n6 = [
    "loyer_m2_median_n6",
    "nb_log_n6",
    "taux_rendement_n6"
]

df[colonnes_n6].isna().nunique(axis=1).value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
#Visualisation des variables catégorielles
fig, axes = plt.subplots(4, 4, figsize=(20, 16))
axes = axes.flatten()

for i, col in enumerate(df_str.columns[:16]):
    sns.countplot(
        data=df_str,
        y=col,
        order=df_str[col].value_counts().index,
        ax=axes[i]
    )
    axes[i].set_title(col)
    axes[i].tick_params(axis="x", rotation=45)

# Masquer les emplacements inutilisés
for ax in axes[len(df_str.columns[:16]):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
#Visualisation des variables quantitatives discrètes
fig, axes = plt.subplots(2, 5, figsize=(20, 12))
axes = axes.flatten()

for i, col in enumerate(colonnes_quanti_d):
    sns.countplot(
        data=df,
        y=col,
        order=df[col].value_counts().sort_index().index,
        ax=axes[i]
    )
    axes[i].set_title(col)
    axes[i].tick_params(axis="x", rotation=45)

# Masquer les 6 graphiques inutilisés
for ax in axes[len(colonnes_quanti_d):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# Variables quantitatives restantes
colonnes_quanti_c = df_quanti.columns.difference(colonnes_quanti_d, sort=False)

fig, axes = plt.subplots(4, 5, figsize=(22, 16))
axes = axes.flatten()

for i, col in enumerate(colonnes_quanti_c[:20]):
    sns.histplot(
        data=df_quanti,
        x=col,
        bins=30,
        ax=axes[i]
    )
    axes[i].set_title(col)

# Masquer les graphiques inutilisés
for ax in axes[len(colonnes_quanti_c[:20]):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4, 5, figsize=(22, 16))
axes = axes.flatten()

for i, col in enumerate(colonnes_quanti_c):
    sns.boxplot(
        data=df_quanti,
        x=col,
        ax=axes[i]
    )
    axes[i].set_title(col)

# Masquer les 2 emplacements inutilisés
for ax in axes[len(colonnes_quanti_c):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

**Constat :**

**Valeurs manquentes**
La comparaison des taux de valeurs manquantes révèle plusieurs profils :
Absences fortement associées au type de bien : surface_terrain (97,2 % pour les appartements contre 20,6 % pour les maisons), charges_copro (52,8 % contre 98,8 %) et ascenseur (36,6 % contre 100 %). Indicateurs agrégés n6 et n7 : les taux de valeurs manquantes sont identiques au sein de chaque groupe de trois variables, mais sensiblement différents entre maisons et appartements. Cela suggère une origine commune des données et une couverture différenciée à investiguer.
Focus sur annee_construction :
65,9 % de valeurs manquantes sur l'ensemble du dataset.
68,0 % pour les appartements et 63,8 % pour les maisons.
61,0 % pour les logements anciens (logement_neuf = n).
90,9 % pour les logements neufs (logement_neuf = o).
L'année de construction est donc particulièrement peu renseignée pour les logements neufs, mais l'absence reste également majoritaire pour les logements anciens. Le caractère neuf du logement est associé à des différences de complétude, sans expliquer à lui seul l'ensemble des valeurs manquantes.

La répartition des annonces est fortement déséquilibrée : la catégorie a représente 75,87 % de l'échantillon, suivie de cm (10,92 %), m (8,26 %), ca (3,49 %) et network (1,32 %). La catégorie b ne représente que 39 annonces (0,14 %) et n'est pas retenue pour l'interprétation comparative.
Les taux de valeurs manquantes présentent des différences importantes entre catégories.
cm (3 028 annonces) présente une absence systématique sur de nombreuses variables descriptives, notamment annee_construction, dpeC et chauffage_energie.
ca et network présentent également des taux élevés d'absence sur plusieurs caractéristiques, avec des profils non identiques.
m présente une meilleure complétude sur plusieurs variables, notamment annee_construction, dpeC et ges_class.
